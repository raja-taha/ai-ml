## TASK : how to create matrices and perform common matrix operations using built-in NumPy methods.

```python
# import required library
import numpy as np
import time
```

```python
[1]*5
```

```python
# demo of speed of python loop vs numpy builtins using vector addition
# Create Python lists
n = 10000000
a = [1]*n
b = [2]*n

start = time.time()

c = [0] * n
for i in range(n):
    c[i] = a[i] + b[i]

end = time.time()

print("Python loop time:", end - start)

# Create NumPy arrays
a_np = np.arange(n)
b_np = np.arange(n)

start = time.time()

c_np = a_np + b_np   # vectorized operation

end = time.time()

print("NumPy vectorized time:", end - start)
```

```python
# create 2 matrices using numpy array
A = np.array([[1,2,3],[0,1,0],[0,0,1]])
B = np.array([[0,1,0],[1,0,0],[1,2,1]])
```

```python
print(A)
print(A.shape)
```

```python
print(B)
print(B.shape)
```

```python
# find sum of the matrices
print(A + B)
```

```python
# find difference of the matrices
print(A-B)
```

```python
# finding product of 2 matrices
print(A@B)
print(np.matmul(A,B))
```

```python
# finding transpose of a matrix
print(A.T) #property
print(np.transpose(A)) #method
```

```python
# finding determinant of a matrix
np.linalg.det(A)
```

```python
# finding inverse of a matrix
np.linalg.inv(A)
```

```python
# finding trace of a matrix
np.trace(A)
```

```python
# finding rank of a matrix
print(np.linalg.matrix_rank(A))
```

```python
# finding eigen value & Eigen vector of a matrix
vals, vecs = np.linalg.eig(B)
print(vals)
print(vecs)
```
