### What is Anaconda?

**Anaconda** is a free, open-source distribution of Python (and R) designed specifically for data science, machine learning, and scientific computing. Think of it as a "starter pack" — instead of installing Python and then manually hunting down dozens of libraries one by one, Anaconda gives you everything bundled together in a single installer.

When you install Anaconda, you get:

- **Python** — the programming language itself.
- **conda** — a powerful package manager *and* environment manager (more on this below).
- **1,500+ pre-installed packages** — popular libraries like NumPy, Pandas, Matplotlib, Scikit-learn, Jupyter, and many more come ready to use out of the box.
- **Anaconda Navigator** — a graphical interface to launch tools (Jupyter Notebook, VS Code, Spyder, etc.) without touching the terminal.

---

### What is Conda?

**Conda** is the command-line tool that comes with Anaconda. It does two jobs:

1. **Package manager** — installs, updates, and removes libraries (similar to `pip`, but it can also handle non-Python dependencies like C libraries and CUDA toolkits).
2. **Environment manager** — creates isolated "virtual environments" so each project can have its own Python version and its own set of libraries, without interfering with other projects.

---

### Why do we need it?

Imagine you are working on two projects:

- **Project A** needs Python 3.9 and TensorFlow 2.10
- **Project B** needs Python 3.11 and TensorFlow 2.15

If you install everything globally (system-wide), these two projects will conflict — upgrading TensorFlow for Project B might break Project A. This is called **dependency hell**.

**Conda environments solve this problem.** Each environment is a completely separate sandbox:

| | Project A env | Project B env |
|---|---|---|
| Python | 3.9 | 3.11 |
| TensorFlow | 2.10 | 2.15 |
| Other libs | specific versions | different versions |

You switch between them with a single command (`conda activate project_a`), and they never interfere with each other.

**In short:**

- **Without Anaconda** — you install Python manually, install packages one by one with `pip`, and risk breaking things when projects need different versions.
- **With Anaconda** — you get Python + all common data science libraries in one click, and you can create isolated environments for each project so nothing ever conflicts.

---

### Anaconda vs Miniconda — Which one?

| | Anaconda | Miniconda |
|---|---|---|
| Size | ~4.5 GB (includes 1,500+ packages) | ~80 MB (minimal — only Python + conda) |
| Best for | Beginners who want everything ready | Advanced users who want to install only what they need |
| Packages | Pre-installed | Install manually with `conda install` |

If you are just starting out, **Anaconda** is recommended — it saves time and avoids setup headaches.

A **conda environment** isolates your project's Python version and dependencies from other projects, preventing version conflicts. Conda can manage both Python packages **and** non-Python dependencies (C libraries, CUDA, etc.).

### Prerequisites — Install Anaconda

If you do not have Anaconda installed, follow these steps:

**1. Download Anaconda**

Go to the official Anaconda download page:
[https://www.anaconda.com/download](https://www.anaconda.com/download)

Choose the installer for your operating system:

| OS | Installer |
|---|---|
| **macOS** | `.pkg` graphical installer (recommended) or command-line `.sh` installer |
| **Windows** | `.exe` graphical installer |
| **Linux** | `.sh` command-line installer |

**2. Install Anaconda**

- **macOS (.pkg):** Double-click the downloaded `.pkg` file and follow the on-screen prompts. Install for "Just Me" (recommended).
- **macOS (.sh):** Open Terminal and run:
  ```bash
  bash ~/Downloads/Anaconda3-<version>-MacOSX-x86_64.sh
  ```
  Follow the prompts and type `yes` to accept the license and initialize conda.

- **Windows:** Double-click the `.exe` installer. Choose "Just Me", accept defaults, and check "Add Anaconda to my PATH" if prompted (or use Anaconda Prompt instead).

- **Linux:** Open a terminal and run:
  ```bash
  bash ~/Downloads/Anaconda3-<version>-Linux-x86_64.sh
  ```
  Follow the prompts, accept the license, and allow the installer to initialize conda.

**3. Verify the installation**

Close and reopen your terminal, then run:
```bash
conda --version
```
You should see something like `conda 24.x.x`.

**4. (Optional) Update Conda**
```bash
conda update conda -y
```

#### Step 2 — Create the Conda environment
This creates a new isolated environment named `dam` with a specific Python version.
> **Note:** Replace `dam` with your preferred environment name and `3.11` with your desired Python version.

````python
!conda create -n dam python=3.11 -y
````

#### Step 3 — Activate the Conda environment

Run the following command **in your terminal**:

```bash
conda activate dam
```

> **Important:** `conda activate` cannot be run from inside a notebook cell. Use your **terminal** or select the kernel in VS Code.

> **Tip (VS Code):** Press **Cmd+Shift+P** then select **Python: Select Interpreter** and choose the `dam` conda environment. Then restart the kernel so this notebook uses it.
